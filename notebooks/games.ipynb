# Match Data Documentation

This document provides a description of the dataset containing information about football matches. Each row in the dataset represents a specific match, capturing various details relevant to that match.

## Column Descriptions

 Column Name                         Description                                                                                       
 `game_id`: A unique identifier for each match in the dataset.                                              
 `competition_id`: A unique identifier for the competition in which the match is held.                             
 `season`: The season in which the match took place                                    
 `round`: The round of the competition during which the match occurred  
 `date`: The date when the match was played.                                                    
 `home_club_id`: A unique identifier for the home club participating in the match.                               
 `away_club_id`: A unique identifier for the away club participating in the match.                               
 `home_club_goals`: The number of goals scored by the home club in the match.                                       
 `away_club_goals`: The number of goals scored by the away club in the match.                                       
 `home_club_position`: The position of the home club in the league table at the time of the match.                     
 `away_club_position`: The position of the away club in the league table at the time of the match.                     
 `home_club_manager_name`: The name of the manager of the home club during the match.                                      
 `away_club_manager_name`: The name of the manager of the away club during the match.                                      
 `stadium`: The name of the stadium where the match took place.                                             
 `attendance`: The number of spectators present at the match.                                                  
 `referee`: The name of the referee officiating the match.                                                  
 `url`: A link to additional information or statistics about the match.                                   
 `home_club_formation`: The formation used by the home club during the match.                  
 `away_club_formation`: The formation used by the away club during the match.                                          
 `home_club_name`: The name of the home club participating in the match.                                           
 `away_club_name`: The name of the away club participating in the match.                                           
 `aggregate`: The aggregate score of the match if applicable                   
 `competition_type`: The type of competition
