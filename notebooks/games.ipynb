## Column Descriptions

```python
# Check if the format of the date column is datetime or object, it's converted to datetime if it's in object format to be able to work with it
if df['date'].dtype == 'object':
    # Convert the date column to datetime format
    df['date'] = pd.to_datetime(df['date'])
    print('Date column converted to datetime format')
else:
    print('Date column is already in datetime format')
    
df['competition_type'] = df['competition_type'].str.upper()
df['competition_type'] = df['competition_type'].str.replace('_', ' ')
df
```

```python
fig, axs = plt.subplots(1, 2, figsize=(18, 8), dpi=80)

# Calculate the percentage of missing values for each column
missing_data = df.isnull().mean() * 100

# Separate columns with missing values greater than 2% and those less than or equal to 2%
filtered_missing_data = missing_data[missing_data > 2].sort_values(ascending=True)
other_missing_data = missing_data[missing_data <= 2].sum()

# Add the "other" category for columns with missing values <= 2%
filtered_missing_data['other'] = other_missing_data

# Sort the data in ascending order to display the missing data in a more readable format
filtered_missing_data.sort_values(ascending=True, inplace=True)

# Pie chart for missing data
axs[0].pie(filtered_missing_data, labels=filtered_missing_data.index, autopct='%1.1f%%', startangle=90, 
           colors=sns.color_palette('viridis', len(filtered_missing_data)))

# Add a center circle for the donut chart
center_circle = plt.Circle((0, 0), 0.70, fc='white')
axs[0].add_artist(center_circle)

# Horizontal bar chart for missing data greater than 2%
axs[1].barh(filtered_missing_data.index, filtered_missing_data.values, color=sns.color_palette('viridis', len(filtered_missing_data)))
axs[1].set_title('Percentage of Missing Values in Each Column')
axs[1].set_xlabel('Percentage')

# Adjust layout to prevent overlap
plt.tight_layout()

# Show the plot
plt.show()
```

1. **Left (Donut Chart)**:

   - The larger sections correspond to columns with a significant percentage of missing data, the big ones are `home_club_formation`, `away_club_formation` in the same percentage, same happens for `home_club_position`-`away_club_position` and `home_club_name`-`away_club_name`.

2. **Right (Horizontal Bar Chart)**:

```python
# Check for competitions in which the home club scored more than 10 goals
more_than_10_home = df.query('home_club_goals > 10')
more_than_10_home = more_than_10_home[['competition_id','season']]

# Check for competitions in which the away club scored more than 10 goals
more_than_10_guest = df.query('away_club_goals > 10')
more_than_10_guest = more_than_10_guest[['competition_id','season']]

# Count occurrences of each competition for home goals
home_goals_counts = more_than_10_home['competition_id'].value_counts().reset_index()
home_goals_counts.columns = ['competition_id', 'home_goals_count']

# Count occurrences of each competition for away goals
away_goals_counts = more_than_10_guest['competition_id'].value_counts().reset_index()
away_goals_counts.columns = ['competition_id', 'away_goals_count']

# Merge the two counts on competition_id
merged_counts = pd.merge(home_goals_counts, away_goals_counts, on='competition_id', how='outer').fillna(0)

# Set the competition_id as index for better plotting
merged_counts.set_index('competition_id', inplace=True)

# Plotting
plt.figure(figsize=(14, 8))
bar_width = 0.35
x = range(len(merged_counts))

# Bar plots for home and away goals
plt.bar(x, merged_counts['home_goals_count'], width=bar_width, label='Home Goals > 10', color='skyblue')
plt.bar([p + bar_width for p in x], merged_counts['away_goals_count'], width=bar_width, label='Away Goals > 10', color='salmon')

# Labeling the plot
plt.xlabel('Competition ID', fontsize=12)
plt.ylabel('Number of Matches', fontsize=12)
plt.title('Competitions Where Clubs Scored More Than 10 Goals', fontsize=14)
plt.xticks([p + bar_width / 2 for p in x], merged_counts.index, rotation=0)
plt.legend()
plt.grid(axis='y')

# Show plot
plt.tight_layout()
plt.show()
```

```python
import geopandas as gpd

# Load the GeoDataFrame
geo_df = gpd.read_file('../ne_110m_admin_0_countries/ne_110m_admin_0_countries.shp')

# Filter the DataFrame for domestic league competitions
df_domestic = df[df['competition_type'] == 'DOMESTIC LEAGUE']

# Aggregate the guest goals by country for domestic leagues
goals_by_country = df_domestic.groupby('country')['away_club_goals'].sum().reset_index()

# Merge with the GeoDataFrame
merged = geo_df.merge(goals_by_country, left_on='ADMIN', right_on='country', how='left')

# Fill NaN values with 0 for countries with no guest goals
merged['away_club_goals'] = merged['away_club_goals'].fillna(0)

# Filter to include only countries with goals > 0 and create a copy
merged_goals_positive = merged[merged['away_club_goals'] > 0].copy()

# Sort the DataFrame by away_club_goals in descending order
merged_goals_positive.sort_values(by='away_club_goals', ascending=False, inplace=True)

# Create subplots: one for the map and one for the bar chart
fig, (ax_map, ax_bar) = plt.subplots(1, 2, figsize=(16, 8))

# Plot the world map
merged.boundary.plot(ax=ax_map, linewidth=1)
merged.plot(column='away_club_goals', ax=ax_map, legend=True,
            legend_kwds={'label': "Total Guest Goals by Country",
                         'orientation': "horizontal"},
            cmap='OrRd', missing_kwds={"color": "lightgrey"})

ax_map.set_title('World Map Colored by Total Guest Goals (Domestic Leagues)')
ax_map.set_xlabel('Longitude')
ax_map.set_ylabel('Latitude')

# Plot the bar chart
ax_bar.barh(merged_goals_positive['country'], merged_goals_positive['away_club_goals'], color='steelblue')
ax_bar.set_title('Countries by Total Guest Goals (Domestic Leagues)')
ax_bar.set_xlabel('Total Guest Goals')
ax_bar.set_ylabel('Country')

# Adjust layout
plt.tight_layout()

# Show the plots
plt.show()
```

Let's see how many domestic competitions are in the dataset and what's the mean of each competition's guest goals



```python
# Filter the DataFrame to include only domestic competitions
domestic_competitions_df = df[df['competition_type'] == 'DOMESTIC LEAGUE']

# Count the number of unique game_id for each country and the mean of away_club_goals for each competition
competitions_summary = domestic_competitions_df.groupby('country').agg(
    unique_game_count=('game_id', 'nunique'),          # Count unique game_id
    mean_guest_goals=('away_club_goals', 'mean'),     # Calculate mean guest goals
    mean_home_goals=('home_club_goals', 'mean')        # Calculate mean home goals
).reset_index()

# Sort by unique_game_count in descending order
competitions_summary.sort_values(by='unique_game_count', ascending=False, inplace=True)

# Create new columns for hypothetical total goals by multiplying unique_game_count and mean_guest_goals
competitions_summary['hypothetical_total_guest_goals'] = competitions_summary['unique_game_count'] * competitions_summary['mean_guest_goals']
competitions_summary['hypothetical_total_home_goals'] = competitions_summary['unique_game_count'] * competitions_summary['mean_home_goals']

difference_home_guest = competitions_summary['hypothetical_total_home_goals'] - competitions_summary['hypothetical_total_guest_goals']

competitions_summary['difference_home_guest'] = difference_home_guest
print(competitions_summary['difference_home_guest'].mean())
# Display the result
competitions_summary
```

```python
import matplotlib.pyplot as plt
import matplotlib as mpl

# Filter the DataFrame to include only domestic competitions
domestic_competitions_df = df[df['competition_type'] == 'DOMESTIC LEAGUE']

# Group by season and country, count the number of unique games
games_per_year_country = domestic_competitions_df.groupby(['season', 'country'])['game_id'].nunique().reset_index()

# Ensure seasons are sorted correctly for plotting
games_per_year_country['season'] = pd.to_datetime(games_per_year_country['season'], format='%Y')
games_per_year_country.sort_values('season', inplace=True)

# Create a line plot
plt.figure(figsize=(20, 10))

# Get a colormap with as many colors as there are unique countries
cmap = mpl.colormaps['tab20b']  # Access the colormap
num_countries = len(games_per_year_country['country'].unique())

# Create a line for each country with a unique color
for i, country in enumerate(games_per_year_country['country'].unique()):
    country_data = games_per_year_country[games_per_year_country['country'] == country]
    # Add the label here
    plt.plot(country_data['season'], country_data['game_id'], marker='x', label=country, color=cmap(i % num_countries))

# Add details to the plot
plt.title('Trends in the Number of Games Played by Country')
plt.xlabel('Year')
plt.ylabel('Number of Games Played')
plt.xticks(rotation=45)  # Rotate x-axis labels for better visibility
plt.legend(title='Country', bbox_to_anchor=(1.05, 1), loc='upper left')  # Legend should now work properly
plt.grid()
plt.tight_layout()

# Show the plot
plt.show()
```
